```python
#11 modificaciones
atletas["n_pruebas"] = atletas.groupby(["ID", "Games"])["Event"].transform("count")

def clasificar_grupo(n):
    if n == 1:
        return "Grupo 1"
    elif n <= 3:
        return "Grupo 2"
    else:
        return "Grupo 3"

atletas["grupo_participacion"] = atletas["n_pruebas"].map(clasificar_grupo)

print("Cantidad de registros por grupo:")
print(atletas["grupo_participacion"].value_counts())

print("\nDeportes principales por grupo:")
display(pd.crosstab(atletas["Sport"], atletas["grupo_participacion"]))

for g in ["Grupo 1", "Grupo 2", "Grupo 3"]:
    print(f"\nTop 10 deportes en {g}:")
    print(atletas[atletas["grupo_participacion"] == g]["Sport"].value_counts().head(10))

print("El Grupo 1 es el mas grande porque la mayoria compite en una sola prueba por edicion")
print("El Grupo 3 es el mas pequeno y se concentra sobre todo en deportes con muchas pruebas individuales como gimnasia, natacion o atletismo")
```
