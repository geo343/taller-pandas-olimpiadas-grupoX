```python
#1
import io
import pandas as pd
from google.colab import files

# raw = pd.read_csv("athlete_events.csv")
# texto_limpio = raw.columns[0] + "\n" + "\n".join(raw.iloc[:, 0])
# df = pd.read_csv(io.StringIO(texto_limpio))

# df.to_csv("athlete_events_fixed.csv", index=False)
# files.download("athlete_events_fixed.csv")

df_atletas = pd.read_csv("athlete_events_fixed.csv")
print('Atletas:----------------------------------------------------------------')
display(df_atletas)

df_regiones = pd.read_csv("noc_regions.csv")
print('Regiones:---------------------------------------------------------------')
display(df_regiones)
```

```python
#7

print("Rango de años:", df_atletas["Year"].min(), "-", df_atletas["Year"].max())

display(df_atletas["Season"].value_counts())

print('\n')


display(df_atletas["Sex"].value_counts())

print('\n')

display(pd.crosstab(df_atletas["Season"], df_atletas["Sex"]))

print('\n')

display(df_atletas.groupby("Season")["Year"].agg(["min", "max", "count"]))

print('\n')

print("Los datos cubren de 1896 a 2016, pero no representan toda la historia olimpica de forma uniforme")
print("Hay mucha mayor participacion en Summer que en Winter y el sexo masculino concentra la mayoria de registros")
print("Ademas los datos llegan hasta 2016, por lo que no incluyen ediciones posteriores")
```

```python
#10

print('A')
display(df_atletas.groupby('Sex')['Medal'].count())

print('No, la verdad los hombres segun datos historicos, consiguieron mas medallas que las mujeres.')

edades_m = df_atletas[df_atletas['Age'] < 18]

print('\n')

print('B')
print('MENORES')
print(edades_m.groupby('Age')['Medal'].count().sum())
edades_M = df_atletas[df_atletas['Age'] >= 18]
print('\n')
print('MAYORES')
print(edades_M.groupby('Age')['Medal'].count().sum())
print('\n')
print('No, la verdad es que las personas con 18 años o mas tienen mas medallas que los jovenes menores de 18 años')
print('\n')

print('C')
menores_18 = df_atletas[df_atletas['Age'] < 18]['Medal'].count()
mayores_35 = df_atletas[df_atletas['Age'] >= 35]['Medal'].count()
intermedios = df_atletas[(df_atletas['Age'] >= 18) & (df_atletas['Age'] < 35)]['Medal'].count()

print('MENORES DE 18:', menores_18)
print('ENTRE 18 Y 34:', intermedios)
print('35 O MAS:', mayores_35)
print('\n')
print('No, la mayoría de medallas las ganó el grupo entre 18 y 34 años, tanto en cantidad como proporcionalmente.')
print('\n')

print('D')
display(df_atletas.groupby('NOC')['Medal'].count().sort_values(ascending=False).head(10))

usa_medals = df_atletas[df_atletas['NOC'] == 'USA']['Medal'].count()
otros_medals = df_atletas[df_atletas['NOC'] != 'USA']['Medal'].count()

print('\n')
print('USA:', usa_medals)
print('OTROS PAISES:', otros_medals)
print('\n')
print('No, aunque USA es el país con más medallas individuales, los demás países combinados ganaron la mayor cantidad y proporción.')
print('\n')

print('E')
display(df_atletas.groupby('Season')['Medal'].count())

print('\n')
print('Sí, en los Juegos de Verano se ganaron muchas más medallas que en los de Invierno debido a que tienen más deportes y eventos.')
print('\n')
```

```python
#11
df_atletas["n_pruebas"] = df_atletas.groupby(["ID", "Games"])["Event"].transform("count")

def clasificar_grupo(n):
    if n == 1:
        return "Grupo 1"
    elif n <= 3:
        return "Grupo 2"
    else:
        return "Grupo 3"

df_atletas["grupo_participacion"] = df_atletas["n_pruebas"].map(clasificar_grupo)

print("Cantidad de registros por grupo:")
print(df_atletas["grupo_participacion"].value_counts())

print("\nDeportes principales por grupo:")
display(pd.crosstab(df_atletas["Sport"], df_atletas["grupo_participacion"]))

for g in ["Grupo 1", "Grupo 2", "Grupo 3"]:
    print(f"\nTop 10 deportes en {g}:")
    print(df_atletas[df_atletas["grupo_participacion"] == g]["Sport"].value_counts().head(10))

print("El Grupo 1 es el mas grande porque la mayoria compite en una sola prueba por edicion")
print("El Grupo 3 es el mas pequeno y se concentra sobre todo en deportes con muchas pruebas individuales como gimnasia, natacion o atletismo")
```

```python
#12
df_atletas["grupo_etario"]=df_atletas["Age"].map(lambda x: "menor" if x<18 else "adulto joven" if 18<=x<=34 else "mayor")
display(pd.crosstab(df_atletas["grupo_etario"], df_atletas["Sex"]))
display(df_atletas.isna().groupby("grupo_etario")[["Height","Weight"]].sum())
display(df_atletas.notna().groupby("grupo_etario")[["Height","Weight"]].sum())
display(df_atletas.groupby("grupo_etario")["medalla"].sum())
```
