```python
import pandas as pd

df_raw = pd.read_csv('../data/1.csv', header=None)

df_atletas = df_raw[0].str.split(',', expand=True)

df_atletas = df_atletas.iloc[:, :15]

df_atletas.columns = df_atletas.iloc[0].str.replace('"', '').str.strip()

df_atletas = df_atletas.iloc[1:].reset_index(drop=True)

print(f"Total de registros de atletas: {df_atletas.shape[0]}")

display(df_atletas.head(10))
```

```python
import pandas as pd

df_regiones = pd.read_csv("noc_regions.csv")

print(f"Total de regiones/países cargados: {df_regiones.shape[0]}")
display(df_regiones.head(40))
```

```python
import io
import pandas as pd

cleaned_lines = []
with open("1.csv", "r", encoding="utf-8", errors="ignore") as f:
  for line in f:
    line = line.strip()
    if line.startswith('"') and line.endswith('"'):
      line = line[1:-1]
    line = line.replace('""', '"')
    cleaned_lines.append(line)

df_atletas = pd.read_csv(io.StringIO("\n".join(cleaned_lines)))

print(
    f"-> df_atletas cargado correctamente con {df_atletas.shape[0]} registros y"
    f" {df_atletas.shape[1]} columnas."
)
display(df_atletas.head(40))

df_regiones = pd.read_csv("noc_regions.csv")

print(
    f"-> df_regiones cargado correctamente con {df_regiones.shape[0]} registros"
    f" y {df_regiones.shape[1]} columnas."
)
display(df_regiones.head(40))
```

```python
display(df_atletas.describe(include='all'))
df_atletas.info()

display(df_regiones.describe(include='all'))
df_regiones.info()
```

```python
df_atletas['IMC'] = df_atletas['Weight'] / ((df_atletas['Height'] / 100) ** 2)

display(df_atletas[['Name', 'Sport', 'Height', 'Weight', 'IMC']].head(10))
```

```python
df_atletas = pd.merge(df_atletas, df_regiones, on='NOC', how='left')

print(f"Total de filas en df_atletas después del merge: {df_atletas.shape[0]}")

filas_sin_region = df_atletas['region'].isnull().sum()
print(f"Filas sin región asignada (NaN): {filas_sin_region}")
```

```python
df_atletas['medalla'] = df_atletas['Medal'].notna().astype(int)

total_registros = len(df_atletas)
atletas_unicos = df_atletas['ID'].nunique()

print(f"Número total de filas (participaciones): {total_registros}")
print(f"Número de atletas únicos (por ID): {atletas_unicos}")

display(df_atletas[['Name', 'Sport', 'Medal', 'medalla']].head(10))
```
