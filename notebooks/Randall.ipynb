```python
import pandas as pd

df_raw = pd.read_csv('../data/athlete_events.csv', header=None)

df_atletas = df_raw[0].str.split(',', expand=True)

df_atletas = df_atletas.iloc[:, :15]

df_atletas.columns = df_atletas.iloc[0].str.replace('', '').str.strip()

df_atletas = df_atletas.iloc[1:].reset_index(drop=True)

print(f"Total de registros de atletas: {df_atletas.shape[0]}")

display(df_atletas.head(10))
```

```python
import pandas as pd

df_regiones = pd.read_csv("../data/noc_regions.csv")

print(f"Total de regiones/países cargados: {df_regiones.shape[0]}")
display(df_regiones.head(40))
```

```python
import pandas as pd

def obtener_resumen_variables(df, nombre_df):
    resumen = pd.DataFrame({
        'Columna': df.columns,
        'Tipo de Dato': df.dtypes.astype(str),
        'Valores No Nulos': df.notnull().sum(),
        'Valores Nulos': df.isnull().sum(),
        'Total Filas': len(df)
    })
    print(f"=== Tabla de Variables: {nombre_df} ===")
    display(resumen.reset_index(drop=True))

obtener_resumen_variables(df_atletas, "Atletas")

obtener_resumen_variables(df_regiones, "Regiones")

print("\n=== Estadísticas Básicas: Atletas ===")
display(df_atletas.describe(include='all'))

print("\n=== Estadísticas Básicas: Regiones ===")
display(df_regiones.describe(include='all'))
```

```python
df_atletas.columns = df_atletas.columns.str.replace('"', '').str.strip()

df_atletas['Height'] = pd.to_numeric(df_atletas['Height'], errors='coerce')
df_atletas['Weight'] = pd.to_numeric(df_atletas['Weight'], errors='coerce')

df_atletas['IMC'] = df_atletas['Weight'] / (df_atletas['Height'] / 100) ** 2

display(df_atletas[['Name', 'Sport', 'Height', 'Weight', 'IMC']].head(10))
```

4. 

Me parece una variable útil como primera aproximación, porque resume en un solo número la relación entre peso y estatura. Sin embargo, no significa lo mismo en todos los deportes. En disciplinas de fuerza como la halterofilia o la lucha, el IMC de un atleta puede marcar sobrepeso u obesidad cuando en realidad el peso extra es masa muscular y no grasa. En deportes como la gimnasia o el atletismo de fondo ocurre lo contrario: los cuerpos son más livianos y el mismo valor se interpreta distinto. Por eso, el IMC debe leerse siempre teniendo en cuenta el deporte.

```python
df_completo = pd.merge(df_atletas, df_regiones, on='NOC', how='left')

print(f"Filas originales en atletas: {len(df_atletas)}")
print(f"Filas después del merge: {len(df_completo)}")

filas_sin_region = df_completo['region'].isnull().sum()
print(f"Filas sin región asignada: {filas_sin_region}")

display(df_completo[['Name', 'NOC', 'region', 'Sport']].head(5))
```

5. 

Usé una fusión how='left', con el dataframe de atletas como tabla principal. Esto garantiza que ninguna fila de atletas se pierda y que el número de filas se mantenga igual después de la fusión, que era justo lo que se pedía verificar. Algunas filas quedan sin región porque ciertos códigos NOC de los atletas no aparecen en la tabla de regiones: son códigos de países que ya no existen, códigos antiguos o equipos que no tienen una región asignada. Con el left join esas filas se conservan, solo que con la región vacía.

```python
df_completo['medalla'] = df_completo['Medal'].notnull().astype(int)

total_filas = len(df_completo)
atletas_unicos = df_completo['ID'].nunique()

print(f"\nTotal de registros (filas): {total_filas}")
print(f"Cantidad de atletas únicos (ID): {atletas_unicos}")

display(df_completo[['Name', 'Medal', 'medalla']].head(10))
```

6.

Creé la variable medalla asignando 1 a los registros con oro, plata o bronce, y 0 a los que tenían la columna Medal vacía, ya que un valor faltante aquí significa que el atleta no ganó medalla en esa prueba, no que el dato se perdió. Al comparar el número de filas con el de atletas únicos (ID) se ve que una misma persona aparece varias veces, una por cada prueba en la que participó. Esto afecta el conteo de medallas, sobre todo en deportes de equipo: si un equipo gana una medalla, cada integrante tiene su propia fila con medalla, así que sumar filas directamente infla el total. Para contar medallas reales hay que agrupar por evento y país (o equipo) y contar cada medalla una sola vez.
